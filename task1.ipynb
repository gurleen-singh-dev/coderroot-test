```python
df = pd.read_csv(
    'https://raw.githubusercontent.com/gurmindero7/test_datasets/refs/heads/main/employee_data.csv'
)
df.head()
```

inspecting dataset

```python
df.info()
```

removing duplicate values

```python
df.drop_duplicates(inplace=True)
```

missing values count in each column


```python
missing_values = df.isnull().sum()
missing_values
```

```python
df.dropna(inplace=True)
```

```python
missing_values = df.isnull().sum()
missing_values
```
