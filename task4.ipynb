### Sentiment analysis

```python
import pandas as pd 
import numpy as np
```

importing the dataset

```python
df = pd.read_csv(
    'https://raw.githubusercontent.com/gurmindero7/test_datasets/refs/heads/main/sentiment_analysis.csv'
)
```

```python
df.info()
df.head(5)
```

Cleaning the text data

```python
# Clean the text data (remove stopwords, punctuation, and special characters).
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(df['text'])
```

coverting test data into numerical format using TF-IDF

```python
#coverting test data into numerical format using TF-IDF
from sklearn.feature_extraction.text import TfidfTransformer
tfidf_transformer = TfidfTransformer()
X_tfidf = tfidf_transformer.fit_transform(X)
```
